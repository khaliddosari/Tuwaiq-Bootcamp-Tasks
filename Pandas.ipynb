```python
import pandas as pd
import numpy as np

# ==========================================
# Upload CSV File
# ==========================================

#from google.colab import files

#uploaded = files.upload()
```

```python
df = pd.read_csv("diamonds.csv")

# ==========================================
# Display Original Data
# ==========================================

print("Original Data:")
print(df)


```

```python

# ==========================================
# Basic Exploration
# ==========================================

# First 5 rows
print("First 5 Rows:")
print(df.head())

print("\n" + "="*50)

# Last 5 rows
print("Last 5 Rows:")
print(df.tail())

print("\n" + "="*50)

# Shape
print("Dataset Shape:")
print(df.shape)

print("\n" + "="*50)

# Columns
print("Columns:")
print(df.columns)

print("\n" + "="*50)

# Info
print("Dataset Info:")
print(df.info())

print("\n" + "="*50)

# Data types
print("Data Types:")
print(df.dtypes)

print("\n" + "="*50)

# Statistics
print("Statistics:")
print(df.describe())

print("\n" + "="*50)

# ==========================================
# Data Cleaning
# ==========================================

# Clean column names
df.columns = df.columns.str.strip().str.lower()

print("Columns After Cleaning:")
print(df.columns)

print("\n" + "="*50)

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

print("\n" + "="*50)

# Add missing values for practice
df.loc[2, "salary"] = np.nan
df.loc[4, "performance_rating"] = np.nan

print("After Adding Missing Values:")
print(df)

print("\n" + "="*50)

```

```python
# Fill missing salary with mean
df["salary"] = df["salary"].fillna(df["salary"].mean())

# Fill missing rating with mean
df["performance_rating"] = df["performance_rating"].fillna(
    df["salary"].mean()
)

print("After Filling Missing Values:")
print(df)

print("\n" + "="*50)

# Add duplicate row
df.loc[len(df)] = df.loc[0]

print("After Adding Duplicate Row:")
print(df)

print("\n" + "="*50)

# Check duplicates
print("Duplicate Rows:")
print(df.duplicated().sum())

print("\n" + "="*50)

# Remove duplicates
df = df.drop_duplicates()

print("After Removing Duplicates:")
print(df)

print("\n" + "="*50)
```
